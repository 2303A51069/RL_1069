```python
# Task 1: Greedy Algorithm using Arithmetic Mean

# Take number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Take machine names and rewards
for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by space: ").split()))
    machines[name] = rewards

# Find Q value for each machine
highest_q = float("-inf")
selected_machine = ""

for name, rewards in machines.items():

    # Calculate arithmetic mean
    q = sum(rewards) / len(rewards)

    print("\nMachine", name)
    print("Rewards:", rewards)
    print(f"Q({name}) = {q}")

    # Greedy selection
    if q > highest_q:
        highest_q = q
        selected_machine = name

# Display selected machine
print("\nGreedy Selected Machine:", selected_machine)
```

```python
# Task 2: Greedy Algorithm using Incremental Update

# Take number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Take machine names and rewards
for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by space: ").split()))
    machines[name] = rewards

# Initialize Q and N
Q = {}
N = {}

for name in machines:
    Q[name] = 0
    N[name] = 0

# Update Q after every reward
for name, rewards in machines.items():

    print("\nMachine", name)

    for reward in rewards:

        # Increment number of plays
        N[name] = N[name] + 1

        # Incremental update formula
        Q[name] = Q[name] + (reward - Q[name]) / N[name]

        print(
            f"Reward = {reward}, "
            f"N({name}) = {N[name]}, "
            f"Q({name}) = {Q[name]:.2f}"
        )

# Find machine with highest Q value
selected_machine = max(Q, key=Q.get)

# Display final Q values
print("\nFinal Q Values:")

for name in Q:
    print(f"Q({name}) = {Q[name]:.2f}")

print("\nGreedy Selected Machine:", selected_machine)
```
