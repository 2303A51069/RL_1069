```python
# Greedy Algorithm using Arithmetic Mean

M = int(input("Number of machines in casino: "))

machines = []

# Enter machine names
for m in range(M):
    name = input("Enter name of the machine: ")
    machines.append(name)

print("Machines:", machines)

Q = []

# Calculate average reward for each machine
for machine in machines:

    N = int(input(f"Enter number of plays happened for {machine}: "))

    average_reward = 0

    for play in range(N):
        reward = int(input(f"Enter reward in play {play + 1}: "))
        average_reward = average_reward + reward

    average_reward = average_reward / N

    Q.append(average_reward)

    print(f"Average reward of {machine} is {average_reward}")

# Find machine with highest Q value
id = Q.index(max(Q))

print("Q values:", Q)
print("Greedy Selected Machine:", machines[id])
```

```python
# Greedy Algorithm using Incremental Update Method

M = int(input("Number of machines in casino: "))

machines = []

# Enter machine names
for m in range(M):
    name = input("Enter name of the machine: ")
    machines.append(name)

print("Machines:", machines)

Q_value = []

# Calculate Q value for each machine
for machine in machines:

    plays = int(input(f"Enter number of plays for {machine}: "))

    Q = 0
    N = 0

    for play in range(plays):

        R = int(input(f"Enter reward in play {play + 1}: "))

        # Increment number of plays
        N = N + 1

        # Incremental update formula
        Q = Q + (R - Q) / N

        print(f"After play {play + 1}: Q({machine}) = {Q}")

    Q_value.append(Q)

# Display Q values
print("\nQ Values:", Q_value)

# Find machine with highest Q value
id = Q_value.index(max(Q_value))

print("Greedy Selected Machine:", machines[id])
```
